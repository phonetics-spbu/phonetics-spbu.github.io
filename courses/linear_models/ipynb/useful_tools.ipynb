<h2>FFMPEG</h2>

```python
import os
import subprocess
```

```python
!wget https://github.com/phonetics-spbu/phonetics-spbu.github.io/raw/refs/heads/main/public/courses/linear_models/files/silly_little_hat.mp4
```

```python
VIDEO_PATH = "/content/silly_little_hat.mp4"
AUDIO_WAV  = "/content/silly_little_hat.wav"
```

```python
# Извлекаем аудио через ffmpeg и конвертируем в WAV
subprocess.run([
    "ffmpeg", "-y", "-i", VIDEO_PATH,
    "-vn",                    # без видео
    "-acodec", "pcm_s16le",   # WAV PCM 16-bit
    "-ar", "16000",           # задаем частоту дискретизации 16 кГц (необязательно)
    "-ac", "1",               # один канал (тоже необязательно)
    AUDIO_WAV
], check=True)
```

<h2>Whisper</h2>

```python
import whisper
```

```python
# получаем расшифровку аудио
model = whisper.load_model("base")   # tiny/base/small/medium/large
result = model.transcribe(AUDIO_WAV, language="en", word_timestamps=True)
```

```python
result["text"]
```

```python
for segment in result["segments"]:
    print(f"[{segment['start']:.3f} – {segment['end']:.3f}] {segment['text'].strip()}")
```

<h2>PyDub</h2>

```python
from pydub import AudioSegment
```

```python
audio = AudioSegment.from_wav(AUDIO_WAV)
```

```python
start = 43.760
end = 45.300
```

```python
piece = audio[int(start*1000): int(end*1000)]

piece.export("such_a_lovely_man.wav", format="wav")
```
