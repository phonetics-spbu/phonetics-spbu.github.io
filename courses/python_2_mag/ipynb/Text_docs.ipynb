<h2>XML</h2>

```python
import xml.etree.ElementTree as ET

# Создадим тестовый XML
xml_data = """<?xml version="1.0" encoding="UTF-8"?>
<library>
    <book id="1">
        <title>Python Cookbook</title>
        <author>David Beazley</author>
    </book>
    <book id="2">
        <title>Fluent Python</title>
        <author>Luciano Ramalho</author>
    </book>
</library>"""

# Парсинг из строки
root = ET.fromstring(xml_data)

print("--- Книги из XML ---")
for book in root.findall('book'):
    title = book.find('title').text
    author = book.find('author').text
    print(f"[{book.get('id')}] {title} — {author}")
```

```python
tree = ET.parse("III.B.13-CHRIST.XML")
root = tree.getroot()
```

```python
def get_author(rec):
    """Автор на русском: SUBFIELD.R, иначе делаем A+B+G."""
    f = rec.find("FIELD.700") or rec.find("FIELD.701")
    if f is None:
        return "—"
    subs = {s.tag.split(".")[-1]: s.text for s in f}
    if "R" in subs:                      # русская форма, если есть
        return subs["R"]
    parts = [subs.get(k, "") for k in ("A", "B", "G")]
    return " ".join(p for p in parts if p)

def get_title(rec):
    f = rec.find("FIELD.200")
    if f is None:
        return "—"
    for s in f:
        if s.tag.endswith(".A"):
            return s.text
    return "—"

print("--- Книги из XML ---")
for i, rec in enumerate(root.findall("record"), 1):
    author = get_author(rec)
    title = get_title(rec)
    print(f"[{i}] {author} — {title}")
```

<h2>Markdown</h2>

```python
import xml.etree.ElementTree as ET

# Пример XML-структуры (или берем реальный файл и ET.parse('file.xml').getroot())
xml_data = """<note>
    <title>Моя заметка</title>
    <tags>
        <tag>python</tag>
        <tag>obsidian</tag>
    </tags>
    <content>
        <section heading="Введение">
            <p>Это текст с ссылкой на <link target="Другая заметка">другую заметку</link>.</p>
        </section>
    </content>
</note>"""

root = ET.fromstring(xml_data)

# --- Формируем Markdown ---
md_lines = []

# 1. Frontmatter (YAML) с тегами
title = root.find('title').text
tags = [t.text for t in root.findall('tags/tag')]
md_lines.append("---")
md_lines.append(f"title: {title}")
md_lines.append(f"tags: [{', '.join(tags)}]") # или списком ниже
md_lines.append("---")
md_lines.append("") # пустая строка

# 2. Контент
for section in root.findall('content/section'):
    # Заголовок
    heading = section.get('heading')
    md_lines.append(f"## {heading}")
    md_lines.append("")

    for p in section.findall('p'):
        # Обрабатываем текст и ссылки внутри параграфа
        text_content = p.text or ""
        for link in p.findall('link'):
            target = link.get('target')
            link_text = link.text
            # Генерируем ссылку
            text_content += f" [[{target}|{link_text}]]"
        md_lines.append(text_content)
        md_lines.append("")

# Записываем в файл
with open("note.md", "w", encoding="utf-8") as f:
    f.write("\n".join(md_lines))
```

```python
def get_author(rec):
    """Автор на русском: SUBFIELD.R, иначе делаем A+B+G."""
    f = rec.find("FIELD.700") or rec.find("FIELD.701")
    if f is None:
        return "—"
    subs = {s.tag.split(".")[-1]: s.text for s in f}
    if "R" in subs:                      # русская форма, если есть
        return subs["R"]
    parts = [subs.get(k, "") for k in ("A", "B", "G")]
    return " ".join(p for p in parts if p)
```

```python
def get_title(rec):
    f = rec.find("FIELD.200")
    if f is None:
        return "—"
    for s in f:
        if s.tag.endswith(".A"):
            return s.text
    return "—"
```

```python
def get_year(rec):
    f = rec.find("FIELD.210")
    if f is None:
        return "—"
    for s in f:
        if s.tag.endswith(".D"):
            return s.text
    return "—"
```

```python
def get_isbn(rec):
    f = rec.find("FIELD.10")
    if f is None:
        return "—"
    for s in f:
        if s.tag.endswith(".A"):
            return s.text
    return "—"
```

```python
def get_tags(rec):
    f = rec.find("FIELD.606")
    tags = []
    if f is None:
        return tags
    for s in f:
        tags.append(s.text) 
    return tags
```

```python
def xml_to_dict(filename):
    tree = ET.parse(filename)
    root = tree.getroot()

    library = []

    
    for i, rec in enumerate(root.findall("record"), 1):
        dict_xml = {
                "title": get_title(rec),
                "author": get_author(rec),
                "year": get_year(rec),
                "isbn": get_isbn(rec),
                "tags": get_tags(rec)
            }
        library.append(dict_xml)

    return library
```

```python
import os

def dict_to_markdown(xml_dict_record, path):
    md_lines = []

    md_lines.append("---")
    md_lines.append(f"title: {xml_dict_record['title']}")
    md_lines.append(f"author: {xml_dict_record['author']}")
    md_lines.append(f"year: {xml_dict_record['year']}")
    md_lines.append(f"isbn: {xml_dict_record['isbn']}")
    md_lines.append("tags:")
    for tag in xml_dict_record["tags"]:
        md_lines.append(f"  - {tag}")
    md_lines.append("---")
    md_lines.append("") # пустая строка

    print(os.path.join(path, f"{xml_dict_record['title'][:15]}.md"))

    with open(os.path.join(path, f"{xml_dict_record['isbn']}.md"), "w", encoding="utf-8") as f:
        f.write("\n".join(md_lines))
```

```python
xml_dict = xml_to_dict("D:\Downloads\III.B.13-CHRIST.XML")
path = "D:\Downloads\md"

for record in xml_dict:
    dict_to_markdown(record, path)
```
