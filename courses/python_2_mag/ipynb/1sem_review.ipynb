<p>Напишите функцию, создающую новый файл только со строками, содержащими определенное слово.</p>

```python
def get_lines_with_word(word, filename):
  with open(filename, "r") as file:
    lines = file.readlines()

  new_lines = []
  for line in lines:
    if word in line.lower().split():
      new_lines.append(line)

  with open("new_file.txt", "w") as file:
    # for line in new_lines:
    file.writelines(new_lines)
```

Задача 4.

Откройте файл 1.txt и прочитайте из него текст. Уберите из него точки, приведите к нижнему регистру и создайте список всех слов, которые начинаются на букву "с". Выведите на экран все эти слова через нижнее подчёркнивание.

```python
with open("1.txt", "r") as file:
    lines = file.readlines()

text = "".join(lines).replace(".", "").lower()
words = text.split()

filtered_list = filter(lambda x: x.startswith("с"), words)
print(set(filtered_list))
```

Задача 5.

а) Дан текст с числами, разделёнными дефисами и пробелами. Написать код, который выводит все числа в возрастающем порядке. Используйте сортировку.

б) Запишите числа в текстовый файл так, чтобы на первой строке были числа от 0 до 9, на второй — от 10 до 99 и так далее.

```python
text = [
    "8-32 4-535-6 10-3 3 30-31-8 32-2 3 5\n",
    "342-12 13789 43 234 04-34 2-1-1-0\n",
    "32-5-31-45-1 34 36 54-123-62-998\n",
]
```

```python
def my_func(number):
  return(int(number))

numbers = "".join(text).replace("-", " ").split()
sorted_numbers = sorted(numbers, key=my_func)
print(sorted_numbers)

last_number = sorted_numbers[-1]
number_of_lines = len(last_number)

lines = [[] for i in range(number_of_lines)]
for number in sorted_numbers:
  number_ = str(int(number))
  lines[len(number_) - 1].append(number)

print(lines)
lines_for_file = []

for line in lines:
  lines_for_file.append(" ".join(line))
  lines_for_file.append("\n")

print(lines_for_file)
with open("numbers.txt", "w") as file:
  file.writelines(lines_for_file)
```
