<h2>CLTK - Classical Language Toolkit</h2>

В версии 1.0 помимо древнегреческого ("grc") и латыни ("lat"), CLTK поддерживает аккадский ("akk"), классический арабский ("arb"), арамейский ("arc"), классический китайский ("lzh"), коптский ("cop"), готский ("got"), хинди ("hin"), средневерхненемецкий ("gmh"), древне- ("ang") и среднеанглийский ("enm"), старо- ("fro") и среднефранцузский ("frm"), церковнославянский ("chu"), древнескандинавский ("non"), пали ("pli"), панджаби ("pan") и санскрит ("san").

Начиная с версии 2.0 CLTK использует существующие модели бэкенда - stanza, openai или ollama. Openai требует ключ, поэтому для примера возьмем бэкенд stanza

```python
from cltk import NLP
from collections import Counter

nlp_lat = NLP("lati1261", backend="stanza", suppress_banner=True)
nlp_grc = NLP("anci1242", backend="stanza", suppress_banner=True)
nlp_chu = NLP("chur1257", backend="stanza", suppress_banner=True)
```

Лемматизация

```python
text_lat = """
Mons parturibat, gemitus immanes ciens,
eratque in terris maxima expectatio.
At ille murem peperit. Hoc scriptum est tibi,
qui, magna cum minaris, extricas nihil.
"""

doc_lat = nlp_lat.analyze(text=text_lat)

# Атрибут upos — это Universal Dependencies POS-тег
for word in doc_lat.words[:15]:
    print(f"{word.string:15} -> {word.lemma:15} [{word.upos}]")
```

```python
text_grc = """Κόραξ τυρὸν ἔχων ἐπí τινος δένδρου ἐκάθητο.
Ἐλθοῦσα δὲ ἀλώπηξ θεασαμένη αὐτὸν τῆς ῥίνης εἴχετο."""

doc_grc = nlp_grc.analyze(text=text_grc)

for word in doc_grc.words:
    print(f"{word.string:15} -> {word.lemma:15} [{word.upos}]")
```

```python
from cltk import NLP

cltk_nlp_chu = NLP("chu")

text = "Въ начало створи Богъ небо и землю въ первый день. И во второй день створи твердь, яже есть посреди воды. Сего же дни раздълишася воды над твердъ, а полъ ихъ под твердъ. А въ 3-й день створи море и ръки источника и съмена. В 4-й день солнце, и луну, и звъзды, и украси Богъ небо…"
cltk_doc = cltk_nlp_chu.analyze(text=text)

for word in cltk_doc.words[:15]:
    print(f"Слово: {word.string:12} -> Лемма: {word.lemma}")
```

Частотный анализ

```python
from collections import Counter

lemmas = [word.lemma for word in doc_lat.words if word.upos.tag != "PUNCT"]
freq = Counter(lemmas)

print("Топ-10 лемм:")
for lemma, count in freq.most_common(10):
    print(f"{lemma}: {count}")
```

Морфологические признаки

```python
# Детальная информация о конкретном слове
word = doc_lat.words[1]

print(f"Слово: {word.string}")
print(f"Лемма: {word.lemma}")
print(f"POS (UD): {word.upos}")
print(f"Признаки: {word.features}")
print(f"Список признаков: {word.features.features}")
print("Извлечение конкретного признака", word.features.features[0].key, word.features.features[0].value, word.features.features[0].value_label)
```

Частотность падежей

```python
doc_grc = nlp_grc.analyze(text=text_grc)

case_counter = Counter()
total_nominal_words = 0

for word in doc_grc.words:
    # Проверяем, что word.features существует и не пустой
    if word.features:
        # Итерируемся по каждому UDFeatureTag в наборе UDFeatureTagSet
        for feature in word.features.features:

            # Проверяем, есть ли у конкретного тега атрибут 'Case'
            if feature.key == "Case":
                case_value = feature.value
                case_counter[case_value] += 1
                total_nominal_words += 1
                break  # У одного слова может быть только один падеж

# Перевод тегов для красивого вывода
case_translation = {
    "Nom": "Именительный (Nominative)",
    "Gen": "Родительный (Genitive)",
    "Dat": "Дательный (Dative)",
    "Acc": "Винительный (Accusative)",
    "Voc": "Звательный (Vocative)"
}

# 4. Вывод результатов статистики
print(f"Всего слов с падежами: {total_nominal_words}\n")
print(f"{'Падеж'} | {'Количество'} | {'Доля (%)'}")
print("-" * 60)

for case, count in case_counter.most_common():
    percentage = (count / total_nominal_words) * 100 if total_nominal_words > 0 else 0
    case_name = case_translation.get(case, case)
    print(f"{case_name} | {count} | {percentage:.2f}%")
```

Частотность времен

```python
tenses = []
for word in doc_grc.words:
    if word.features is not None:
        for tag in word.features.features:
            if tag.key == "Tense":
                tenses.append(tag.value_label)
                break

tense_freq = Counter(tenses)

print("Частотность времён:")
for tense, count in tense_freq.most_common():
    print(f"{tense}: {count}")
```

Синтаксические зависимости

```python
print(f"{'Слово':<15} {'Связь':<15} {'Главное (индекс)':<15} {'Главное (слово)':<15}")
print("-" * 60)

for word in doc_lat.words:
    # Получаем индекс головного слова
    gov_idx = word.governor

    # Для корневого слова (ROOT) индекс может быть -1 или ссылаться на себя
    if gov_idx is not None and 0 <= gov_idx < len(doc_lat.words):
        gov_word = doc_lat.words[gov_idx].string
    else:
        gov_word = "ROOT"

    print(f"{word.string:<15} {word.dependency_relation.name:<15} {str(gov_idx):<15} {gov_word:<15}")
```

<h2>Древнегреческие модели SpaCy</h2>
Нативно SpaCy не поддерживает модели древнегреческого, но их можно скачать отдельно, например, из проекта greCy:

https://github.com/jmyerston/greCy

```python
!pip install https://github.com/jmyerston/greCy/releases/download/v3.8/grc_proiel_sm-3.8-py3-none-any.whl
```

```python
import grc_proiel_sm
import spacy

# Load using the imported module directly
nlp = grc_proiel_sm.load()

text = "Κόραξ τυρὸν ἔχων ἐπí τινος δένδρου ἐκάθητο. Ἐλθοῦσα δὲ ἀλώπηξ θεασαμένη αὐτὸν τῆς ῥίνης εἴχετο."
doc = nlp(text)

for token in doc:
    print(f'{token.text}, lemma: {token.lemma_} pos:{token.pos_}')
```

```python
from spacy.matcher import Matcher

matcher = Matcher(nlp.vocab)

# Ищем конструкцию: существительное + глагол
pattern = [
    {"POS": "NOUN"},
    {"POS": "VERB"}
]
matcher.add("ADJ_NOUN", [pattern])

matches = matcher(doc)
for match_id, start, end in matches:
    span = doc[start:end]
    print(f"Найдено: {span.text}")
```
