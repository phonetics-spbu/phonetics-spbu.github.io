```python
import spacy

nlp = spacy.load("ru_core_news_sm")

text = "Студенты-лингвисты изучают иностранные языки в университете."

doc = nlp(text)

print("Токены:")
for token in doc:
    print(f"{token.text:15} | лемма: {token.lemma_:10} | POS: {token.pos_}")
```

NLP = Natural Language Processing, Neuro-Linguistic Programming - это про другое :))

```python
text = "Она очень быстро прочитала интересную ей книгу."

doc = nlp(text)

print(f"{'Токен':12} {'POS':8} {'TAG':10} {'Морфология'}")
print("-" * 60)
for token in doc:
    print(f"{token.text:12} {token.pos_:8} {token.tag_:10} {token.morph}")
```

Морфологические признаки доступны как словарь:

```python
for token in doc:
    if token.pos_ == "NOUN":
        print(token.text, str(token.morph))
        # {'Animacy': 'Inan', 'Case': 'Acc', 'Gender': 'Fem', 'Number': 'Sing'}
```

Синтаксические зависимости

```python
text = "Маленькая девочка читает интересную книгу в парке."

doc = nlp(text)

print(f"{'Токен':12} {'Dep':10} {'Голова':10} {'Объяснение'}")
print("-" * 70)
for token in doc:
    print(f"{token.text:12} {token.dep_:10} {token.head.text:10} {spacy.explain(token.dep_)}")
```

Визуализация

```python
from spacy import displacy

text = "Студент быстро написал курсовую работу по семантике."
doc = nlp(text)

# Визуализация дерева зависимостей
displacy.render(doc, style="dep", jupyter=True, options={"compact": True})
```

Также можно визуализировать именованные сущности (NER) - это полезно для анализа текстов на предмет упоминания людей, организаций, локаций:

```python
text = "Иван Петров учится в МГУ в Москве с 2022 года."
doc = nlp(text)

displacy.render(doc, style="ent", jupyter=True)
```

Именованные сущности (NER)

```python
text = "Пушкин родился в Москве в 1799 году. Он учился в Царскосельском лицее."

doc = nlp(text)

for ent in doc.ents:
    print(f"{ent.text:25} → {ent.label_:10} ({spacy.explain(ent.label_)})")
```

Поиск по шаблонам (Matcher)

```python
from spacy.matcher import Matcher

nlp = spacy.load("ru_core_news_sm")
matcher = Matcher(nlp.vocab)

# Ищем конструкцию: прилагательное + существительное
pattern = [
    {"POS": "ADJ"},
    {"POS": "NOUN"}
]
matcher.add("ADJ_NOUN", [pattern])

text = "Красная площадь, зелёная трава, интересная книга, кот красивый."
doc = nlp(text)

matches = matcher(doc)
for match_id, start, end in matches:
    span = doc[start:end]
    print(f"Найдено: {span.text}")
```

<h3>Частотность частей речи</h3>

```python
from collections import Counter

# Возьмём фрагмент текста
text = """
Языковедение (языкознание, лингвистика) — наука, изучающая языки (в принципе — все существующие, когда-либо существовавшие и могущие возникнуть в будущем),
а тем самым и человеческий язык вообще. Как всякая наука, языковедение возникло в связи с практическими потребностями, но постепенно развилось в сложную
и разветвленную систему дисциплин как теоретического, так и прикладного характера. Внутри теоретического языковедения условно различают частное и общее.
"""

doc = nlp(text)

pos_counts = Counter(token.pos_ for token in doc if not token.is_punct and not token.is_space)

print("Частотность частей речи:")
for pos, count in pos_counts.most_common():
    print(f"{pos:10} {count}")
```
