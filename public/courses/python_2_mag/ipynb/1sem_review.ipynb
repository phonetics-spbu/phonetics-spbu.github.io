<p>Задача 1.</p>
<p>Напишите функцию, заменяющую в файле все буквы на заглавные. Результат должен записаться в новый файл с другим именем.</p>

```python
!wget https://raw.githubusercontent.com/annashell/python_mag/refs/heads/main/txt_files/1.txt
```

```python
import os

def change_letters_to_capital(filename):
    with open(filename, 'r', encoding='utf-8') as file:
        lines = file.readlines()

    path, file = os.path.split(filename)
    print(path, file)
    file_nm, ext = os.path.splitext(file)
    print(file_nm, ext)
    new_filename = os.path.join(path, file_nm + "_modified" + ext)

    with open(new_filename, 'w') as file:
        for line in lines:
            file.write(line.upper() + '\n')
```

```python
change_letters_to_capital("1.txt")
```

<p>Задача 2.</p>
<p>Напишите функцию, объединияющую содержимое двух файлов в один.</p>

```python
!wget https://raw.githubusercontent.com/annashell/python_mag/refs/heads/main/txt_files/1.txt
!wget https://raw.githubusercontent.com/annashell/python_mag/refs/heads/main/txt_files/2.txt
```

```python
def unite_files(filename1, filename2):
    with open(filename1, 'r', encoding="utf-8") as file1:
        lines_1 = file1.readlines()
    with open(filename2, 'r', encoding="utf-8") as file2:
        lines_2 = file2.readlines()

    lines_1.append('\n')
    lines_1.extend(lines_2)

    path, file = os.path.split(filename1)
    new_filename = os.path.join(path, "united.txt")

    with open(new_filename, 'w', encoding='utf-8') as file_united:
        for line in lines_1:
           file_united.write(line)
```

```python
unite_files("1.txt", "2.txt")
```

<p>Задача 3.</p>
<p>Напишите функцию, создающую новый файл только со строками, содержащими определенное слово.</p>

```python
def find_lines_with_word(filename, word):
    with open(filename, "r", encoding="utf-8") as file:
        lines = file.readlines()

    path, file = os.path.split(filename)
    new_filename = os.path.join(path, "filtered.txt")

    with open(new_filename, 'w', encoding='utf-8') as file:
        for line in lines:
            if word in line.split():
                file.write(line)
```

```python
find_lines_with_word("1.txt", "старое")
```
